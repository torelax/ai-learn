# staticmethod

静态方法的好处
- 内存节省
- 性能更快
- 线程安全
- 支持工厂模式等
- ...

```python
import timeit 

class Calculator:
    @staticmethod 
    def add(a, b):
        return a + b 

    def instance_add(self, a, b):
        return a + b 

# 测试调用速度 
print(timeit.timeit(lambda: Calculator.add(1, 2), number=1000000))
print(timeit.timeit(lambda: Calculator().instance_add(1, 2), number=1000000))
```

```python
from abc import abstractmethod, ABC


class base(ABC):
    def __init__(self) -> None:
        print(1)
        pass
    
    @abstractmethod
    def p(self, a):
        print('sdfsdf')

class subcls(base):
    def __init__(self) -> None:
        pass

    def p(self, a):
        super().p(1)
        print(a, a)
subcls().p(1)
```

# 异常捕获
`try except` 的优先级：

(`try-->except-->finally`) 中的**普通语句** --> `finally` 中的 `return` 语句(若有)-->否则--> `try-except` 中的 `return` 语句
> 注意:正常执行Finally中有return语句的话,执行finally中的return后不再执行try的语句了

```python
def example1():
    try:
        print("Try block")
        return "From try"
    except Exception as e:
        print(f"Exception: {e}")
    else:
        print("Else block")
    finally:
        print("Finally block")

print(example1())
```

```python
def example2():
    try:
        print("Try block")
        raise ValueError("An error occurred")
    except ValueError as e:
        print(f"Exception: {e}")
        return "From except"
    else:
        print("Else block")
    finally:
        print("Finally block")

print(example2())
```

```python
def example3():
    try:
        print("Try block")
    except Exception as e:
        print(f"Exception: {e}")
    else:
        print("Else block")
        return "From else"
    finally:
        print("Finally block")

print(example3())
```

```python
def example4():
    try:
        print("Try block")
        return "From try"
    except Exception as e:
        print(f"Exception: {e}")
    else:
        print("Else block")
    finally:
        print("Finally block")
        return "From finally"

print(example4())
```
